# The HBB Sickle Cell Mutation

## Objective

Investigate the mutation in the HBB gene responsible for sickle hemoglobin (HbS)
and trace its effect from DNA sequence to the resulting amino-acid substitution.

## Questions

1. What is the HBB gene?
2. What protein does HBB encode?
3. What mutation produces HbS?
4. How does the mutation change a codon?
5. How does that codon change the β-globin protein?
6. Why is the resulting amino-acid substitution biochemically important?
